KMeans

Centroid based Algo

it uses elbow method to find best k and uses wcss

```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X,y_true=make_blobs(n_samples=500,centers=3,cluster_std=0.60,random_state=42)
df=pd.DataFrame(X,columns=['Feature_1','Feature 2'])
```

```python
df
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

```python
inertia=[]
K_range=range(1,11)
```

```python
for k in K_range:
    Kmeans=KMeans(n_clusters=k,random_state=42)
    Kmeans.fit(X_scaled)
    inertia.append(Kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(K_range,inertia,marker=0)
```

```python
K_meansfinal=KMeans(n_clusters=3,random_state=42)
cluster_labels=K_meansfinal.fit_predict(X_scaled)
cluster_labels
```

```python
df['cluster']=cluster_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature 2'],
                hue=df['cluster'],
                palette='viridis')
```

DB Scan

Density Based spatial Algorithm 


Non Parametric 
epsilon distance

```python
from sklearn.datasets import make_moons
```

```python
X,y_true=make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
df=pd.DataFrame(X,columns=['Feature 1','Feature2'])
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
Kmeans=KMeans(n_clusters=2,random_state=42)
kmeans_labels=Kmeans.fit_predict(X_scaled)
df['kmeans_cluster']=kmeans_labels
```

```python
sns.scatterplot(x=df['Feature 1'],y=df['Feature2'],hue=df['kmeans_cluster'],palette='tab10')
```

```python
dbscan=DBSCAN(eps=0.3,min_samples=5)
dbscan_labels=dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster']=dbscan_labels
sns.scatterplot(x=df['Feature 1'],y=df['Feature2'],
                hue=df['dbscan_cluster'],palette='tab10')
```
