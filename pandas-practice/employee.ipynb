```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [26, 30, 54, 25, 70, 35, 42, 28, 60, 45,
            32, 50, 23, 65, 38, 29, 41, 55, 33, 48],

    'Category': ['A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'A',
                 'B', 'A', 'B', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],

    'Tension': ['L', 'L', 'H', 'L', 'M', 'H', 'M', 'L', 'H', 'M',
                'L', 'H', 'M', 'H', 'L', 'M', 'H', 'L', 'M', 'H'],

    'Salary': [25000, 30000, 55000, 22000, 80000, 40000, 45000, 28000,
               65000, 50000, 32000, 60000, 27000, 75000, 35000,
               42000, 58000, 31000, 48000, 70000],

    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'HR', 'Finance',
                   'IT', 'HR', 'Finance', 'IT', 'HR', 'Finance',
                   'IT', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT'],

    'Gender': ['M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F']
})

```

Calculate average Salary.

```python
avg_salary=df['Salary'].mean()
print(avg_salary)
```

Calculate maximum and min Age.

```python
max_age=df['Age'].max()
min_age=df['Age'].min()
print(max_age)
print(min_age)
```


	Count total employees.

```python
total=len(df)
print(total)
```

Find average Age for each Gender.

```python
avg_age=df.groupby('Gender')['Age'].mean()
print(avg_age)
```

Find average Salary for each Department.

```python
ans=df.groupby('Department')['Salary'].mean()
print(ans)
```

Find minimum Age for each Tension level.

```python
min_age=df.groupby('Tension')['Age'].min()
print(min_age)
```

	Count employees in each Department.
    

```python
cnt=df.groupby('Department').value_counts()
print(cnt)
```

	Find employees whose Salary is greater than 50,000.

```python
ans=df[df['Salary']>50000]
print(ans)
```
